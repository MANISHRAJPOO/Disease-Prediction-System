```python
# ==========================================
# DISEASE PREDICTION SYSTEM
# STEP 3: DATA PREPROCESSING
# ==========================================

import numpy as np
import pandas as pd

print("=" * 60)
print("             STEP 3: DATA PREPROCESSING")
print("=" * 60)


# ==========================================
# PART A — DIABETES DATASET
# ==========================================

print("\n\nDIABETES DATA PREPROCESSING")
print("-" * 60)

# Make a copy so the original dataset remains unchanged
diabetes_clean = diabetes_data.copy()

print("\nOriginal shape:")
print(diabetes_clean.shape)


# ------------------------------------------
# 1. Check duplicate records
# ------------------------------------------

print("\nDuplicate rows:")
print(diabetes_clean.duplicated().sum())

# Remove duplicate rows
diabetes_clean = diabetes_clean.drop_duplicates()

print("Shape after removing duplicates:")
print(diabetes_clean.shape)


# ------------------------------------------
# 2. Check missing values
# ------------------------------------------

print("\nMissing values before cleaning:")
print(diabetes_clean.isnull().sum())


# ------------------------------------------
# 3. Replace invalid zero values
# ------------------------------------------

# In the diabetes dataset, zero is not realistic
# for these medical measurements.

invalid_zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for column in invalid_zero_columns:

    if column in diabetes_clean.columns:

        diabetes_clean[column] = diabetes_clean[column].replace(
            0,
            np.nan
        )


# ------------------------------------------
# 4. Fill missing values with median
# ------------------------------------------

for column in invalid_zero_columns:

    if column in diabetes_clean.columns:

        diabetes_clean[column] = diabetes_clean[column].fillna(
            diabetes_clean[column].median()
        )


print("\nMissing values after cleaning:")
print(diabetes_clean.isnull().sum())


# ------------------------------------------
# 5. Check data types
# ------------------------------------------

print("\nDiabetes data types:")
print(diabetes_clean.dtypes)


# ------------------------------------------
# 6. Diabetes statistical summary
# ------------------------------------------

print("\nDiabetes statistical summary:")
display(diabetes_clean.describe())


# ==========================================
# PART B — HEART DISEASE DATASET
# ==========================================

print("\n\nHEART DISEASE DATA PREPROCESSING")
print("-" * 60)

# Make a copy
heart_clean = heart_data.copy()

print("\nOriginal shape:")
print(heart_clean.shape)


# ------------------------------------------
# 1. Check duplicate records
# ------------------------------------------

print("\nDuplicate rows:")
print(heart_clean.duplicated().sum())

# Remove duplicates
heart_clean = heart_clean.drop_duplicates()

print("Shape after removing duplicates:")
print(heart_clean.shape)


# ------------------------------------------
# 2. Check missing values
# ------------------------------------------

print("\nMissing values before cleaning:")
print(heart_clean.isnull().sum())


# ------------------------------------------
# 3. Fill missing numerical values
# ------------------------------------------

numeric_columns = heart_clean.select_dtypes(
    include=np.number
).columns

for column in numeric_columns:

    if heart_clean[column].isnull().sum() > 0:

        heart_clean[column] = heart_clean[column].fillna(
            heart_clean[column].median()
        )


# ------------------------------------------
# 4. Check missing values again
# ------------------------------------------

print("\nMissing values after cleaning:")
print(heart_clean.isnull().sum())


# ------------------------------------------
# 5. Check data types
# ------------------------------------------

print("\nHeart disease data types:")
print(heart_clean.dtypes)


# ------------------------------------------
# 6. Heart statistical summary
# ------------------------------------------

print("\nHeart disease statistical summary:")
display(heart_clean.describe())


# ==========================================
# PART C — FINAL DATA CHECK
# ==========================================

print("\n\n" + "=" * 60)
print("             PREPROCESSING SUMMARY")
print("=" * 60)

print("\nDiabetes dataset:")
print("Rows:", diabetes_clean.shape[0])
print("Columns:", diabetes_clean.shape[1])

print("\nHeart disease dataset:")
print("Rows:", heart_clean.shape[0])
print("Columns:", heart_clean.shape[1])

print("\nDiabetes missing values:")
print(diabetes_clean.isnull().sum().sum())

print("\nHeart disease missing values:")
print(heart_clean.isnull().sum().sum())

print("\n" + "=" * 60)
print("       STEP 3 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================
# DISEASE PREDICTION SYSTEM
# STEP 4: FEATURE SELECTION & NORMALIZATION
# ==========================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif

print("=" * 60)
print("       STEP 4: FEATURE SELECTION & NORMALIZATION")
print("=" * 60)


# ==========================================================
# PART A — DIABETES DATASET
# ==========================================================

print("\n\nDIABETES FEATURE SELECTION")
print("-" * 60)

# Separate features and target
X_diabetes = diabetes_clean.drop(
    "Outcome",
    axis=1
)

y_diabetes = diabetes_clean["Outcome"]

print("\nOriginal Diabetes Features:")
print(X_diabetes.columns.tolist())


# ----------------------------------------------------------
# Split data into training and testing sets
# ----------------------------------------------------------

X_train_diabetes, X_test_diabetes, y_train_diabetes, y_test_diabetes = train_test_split(
    X_diabetes,
    y_diabetes,
    test_size=0.20,
    random_state=42,
    stratify=y_diabetes
)

print("\nDiabetes training samples:", X_train_diabetes.shape[0])
print("Diabetes testing samples:", X_test_diabetes.shape[0])


# ----------------------------------------------------------
# Normalize / Standardize Diabetes features
# ----------------------------------------------------------

diabetes_scaler = StandardScaler()

X_train_diabetes_scaled = diabetes_scaler.fit_transform(
    X_train_diabetes
)

X_test_diabetes_scaled = diabetes_scaler.transform(
    X_test_diabetes
)


# ----------------------------------------------------------
# Feature Selection
# ----------------------------------------------------------

# Select the 6 best features
diabetes_selector = SelectKBest(
    score_func=f_classif,
    k=6
)

X_train_diabetes_selected = diabetes_selector.fit_transform(
    X_train_diabetes_scaled,
    y_train_diabetes
)

X_test_diabetes_selected = diabetes_selector.transform(
    X_test_diabetes_scaled
)


# Get selected feature names
selected_diabetes_features = X_diabetes.columns[
    diabetes_selector.get_support()
]

print("\nSelected Diabetes Features:")

for feature in selected_diabetes_features:
    print("✓", feature)


# ==========================================================
# PART B — HEART DISEASE DATASET
# ==========================================================

print("\n\nHEART DISEASE FEATURE SELECTION")
print("-" * 60)

# Separate features and target
X_heart = heart_clean.drop(
    "target",
    axis=1
)

y_heart = heart_clean["target"]

print("\nOriginal Heart Disease Features:")
print(X_heart.columns.tolist())


# ----------------------------------------------------------
# Split data into training and testing sets
# ----------------------------------------------------------

X_train_heart, X_test_heart, y_train_heart, y_test_heart = train_test_split(
    X_heart,
    y_heart,
    test_size=0.20,
    random_state=42,
    stratify=y_heart
)

print("\nHeart training samples:", X_train_heart.shape[0])
print("Heart testing samples:", X_test_heart.shape[0])


# ----------------------------------------------------------
# Normalize / Standardize Heart features
# ----------------------------------------------------------

heart_scaler = StandardScaler()

X_train_heart_scaled = heart_scaler.fit_transform(
    X_train_heart
)

X_test_heart_scaled = heart_scaler.transform(
    X_test_heart
)


# ----------------------------------------------------------
# Feature Selection
# ----------------------------------------------------------

# Select the 8 best features
heart_k = min(
    8,
    X_train_heart.shape[1]
)

heart_selector = SelectKBest(
    score_func=f_classif,
    k=heart_k
)

X_train_heart_selected = heart_selector.fit_transform(
    X_train_heart_scaled,
    y_train_heart
)

X_test_heart_selected = heart_selector.transform(
    X_test_heart_scaled
)


# Get selected feature names
selected_heart_features = X_heart.columns[
    heart_selector.get_support()
]

print("\nSelected Heart Disease Features:")

for feature in selected_heart_features:
    print("✓", feature)


# ==========================================================
# PART C — DISPLAY RESULTS
# ==========================================================

print("\n\n" + "=" * 60)
print("              STEP 4 SUMMARY")
print("=" * 60)

print("\nDiabetes:")
print("Original features:", X_diabetes.shape[1])
print("Selected features:", X_train_diabetes_selected.shape[1])

print("\nSelected Diabetes Features:")
print(list(selected_diabetes_features))


print("\nHeart Disease:")
print("Original features:", X_heart.shape[1])
print("Selected features:", X_train_heart_selected.shape[1])

print("\nSelected Heart Disease Features:")
print(list(selected_heart_features))


print("\n" + "=" * 60)
print("       STEP 4 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================
# DISEASE PREDICTION SYSTEM
# STEP 5: TRAIN MACHINE LEARNING MODELS
# ==========================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

print("=" * 60)
print("       STEP 5: TRAIN MACHINE LEARNING MODELS")
print("=" * 60)


# ==========================================================
# PART A — CREATE MODELS
# ==========================================================

# Logistic Regression
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Random Forest
random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

# Support Vector Machine
svm_model = SVC(
    probability=True,
    random_state=42
)


# ==========================================================
# PART B — TRAIN DIABETES MODELS
# ==========================================================

print("\n\nTRAINING DIABETES MODELS")
print("-" * 60)

# Logistic Regression
logistic_model.fit(
    X_train_diabetes_selected,
    y_train_diabetes
)

print("✓ Diabetes Logistic Regression trained")


# Random Forest
random_forest_model.fit(
    X_train_diabetes_selected,
    y_train_diabetes
)

print("✓ Diabetes Random Forest trained")


# SVM
svm_model.fit(
    X_train_diabetes_selected,
    y_train_diabetes
)

print("✓ Diabetes SVM trained")


# ==========================================================
# PART C — SAVE DIABETES MODELS
# ==========================================================

diabetes_models = {
    "Logistic Regression": logistic_model,
    "Random Forest": random_forest_model,
    "SVM": svm_model
}


# ==========================================================
# PART D — TRAIN HEART DISEASE MODELS
# ==========================================================

print("\n\nTRAINING HEART DISEASE MODELS")
print("-" * 60)

# Create separate models for Heart Disease

heart_logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

heart_random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

heart_svm_model = SVC(
    probability=True,
    random_state=42
)


# Logistic Regression
heart_logistic_model.fit(
    X_train_heart_selected,
    y_train_heart
)

print("✓ Heart Disease Logistic Regression trained")


# Random Forest
heart_random_forest_model.fit(
    X_train_heart_selected,
    y_train_heart
)

print("✓ Heart Disease Random Forest trained")


# SVM
heart_svm_model.fit(
    X_train_heart_selected,
    y_train_heart
)

print("✓ Heart Disease SVM trained")


# ==========================================================
# PART E — STORE HEART MODELS
# ==========================================================

heart_models = {
    "Logistic Regression": heart_logistic_model,
    "Random Forest": heart_random_forest_model,
    "SVM": heart_svm_model
}


# ==========================================================
# PART F — DISPLAY MODEL INFORMATION
# ==========================================================

print("\n\n" + "=" * 60)
print("              TRAINING SUMMARY")
print("=" * 60)

print("\nDiabetes Models:")
for model_name in diabetes_models:
    print("✓", model_name)

print("\nHeart Disease Models:")
for model_name in heart_models:
    print("✓", model_name)


print("\n" + "=" * 60)
print("       STEP 5 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
# ==========================================
# DISEASE PREDICTION SYSTEM
# STEP 6: MODEL EVALUATION
# ==========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


print("=" * 60)
print("           STEP 6: MODEL EVALUATION")
print("=" * 60)


# ==========================================================
# PART A — EVALUATE DIABETES MODELS
# ==========================================================

print("\n\nDIABETES MODEL EVALUATION")
print("-" * 60)

diabetes_results = []

for model_name, model in diabetes_models.items():

    # Make predictions
    predictions = model.predict(
        X_test_diabetes_selected
    )

    # Calculate metrics
    accuracy = accuracy_score(
        y_test_diabetes,
        predictions
    )

    precision = precision_score(
        y_test_diabetes,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test_diabetes,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test_diabetes,
        predictions,
        zero_division=0
    )

    # Store results
    diabetes_results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    # Print results
    print("\nModel:", model_name)
    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nClassification Report:")
    print(
        classification_report(
            y_test_diabetes,
            predictions,
            zero_division=0
        )
    )

    # Confusion Matrix
    cm = confusion_matrix(
        y_test_diabetes,
        predictions
    )

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Diabetes", "Diabetes"],
        yticklabels=["No Diabetes", "Diabetes"]
    )

    plt.title(
        f"Diabetes - {model_name}"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.show()


# ==========================================================
# PART B — DIABETES RESULTS TABLE
# ==========================================================

diabetes_results_df = pd.DataFrame(
    diabetes_results
)

print("\n\nDIABETES RESULTS")
print("-" * 60)

display(diabetes_results_df)


# ==========================================================
# PART C — EVALUATE HEART DISEASE MODELS
# ==========================================================

print("\n\nHEART DISEASE MODEL EVALUATION")
print("-" * 60)

heart_results = []

for model_name, model in heart_models.items():

    # Make predictions
    predictions = model.predict(
        X_test_heart_selected
    )

    # Calculate metrics
    accuracy = accuracy_score(
        y_test_heart,
        predictions
    )

    precision = precision_score(
        y_test_heart,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test_heart,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test_heart,
        predictions,
        zero_division=0
    )

    # Store results
    heart_results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    # Print results
    print("\nModel:", model_name)
    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nClassification Report:")
    print(
        classification_report(
            y_test_heart,
            predictions,
            zero_division=0
        )
    )

    # Confusion Matrix
    cm = confusion_matrix(
        y_test_heart,
        predictions
    )

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Heart Disease", "Heart Disease"],
        yticklabels=["No Heart Disease", "Heart Disease"]
    )

    plt.title(
        f"Heart Disease - {model_name}"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.show()


# ==========================================================
# PART D — HEART RESULTS TABLE
# ==========================================================

heart_results_df = pd.DataFrame(
    heart_results
)

print("\n\nHEART DISEASE RESULTS")
print("-" * 60)

display(heart_results_df)


# ==========================================================
# PART E — FIND BEST MODELS
# ==========================================================

best_diabetes_model_name = (
    diabetes_results_df
    .sort_values(
        by="F1 Score",
        ascending=False
    )
    .iloc[0]["Model"]
)

best_heart_model_name = (
    heart_results_df
    .sort_values(
        by="F1 Score",
        ascending=False
    )
    .iloc[0]["Model"]
)


print("\n" + "=" * 60)
print("                 BEST MODELS")
print("=" * 60)

print(
    "\nBest Diabetes Model:",
    best_diabetes_model_name
)

print(
    "Best Heart Disease Model:",
    best_heart_model_name
)


# ==========================================================
# PART F — COMPARISON GRAPH
# ==========================================================

print("\nGenerating model comparison graphs...")


# Diabetes comparison

diabetes_results_df.set_index(
    "Model"
)[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Diabetes Model Performance Comparison"
)

plt.ylabel("Score")
plt.ylim(0, 1)

plt.xticks(rotation=0)

plt.legend(
    loc="lower right"
)

plt.tight_layout()

plt.show()


# Heart Disease comparison

heart_results_df.set_index(
    "Model"
)[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Heart Disease Model Performance Comparison"
)

plt.ylabel("Score")
plt.ylim(0, 1)

plt.xticks(rotation=0)

plt.legend(
    loc="lower right"
)

plt.tight_layout()

plt.show()


# ==========================================================
# COMPLETED
# ==========================================================

print("\n" + "=" * 60)
print("       STEP 6 COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
import pandas as pd
import numpy as np

# 5-Fold Cross-Validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("5-Fold Stratified Cross-Validation created successfully!")
```

```python
diabetes_cv_results = []

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1"
}

for model_name, model in diabetes_models.items():

    scores = cross_validate(
        model,
        X_train_diabetes_selected,
        y_train_diabetes,
        cv=cv,
        scoring=scoring
    )

    diabetes_cv_results.append({
        "Model": model_name,
        "Mean Accuracy": scores["test_accuracy"].mean(),
        "Mean Precision": scores["test_precision"].mean(),
        "Mean Recall": scores["test_recall"].mean(),
        "Mean F1 Score": scores["test_f1"].mean()
    })

diabetes_cv_df = pd.DataFrame(diabetes_cv_results)

print("=" * 70)
print("       DIABETES - 5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 70)

display(diabetes_cv_df.round(4))
```

```python
heart_cv_results = []

for model_name, model in heart_models.items():

    scores = cross_validate(
        model,
        X_train_heart_selected,
        y_train_heart,
        cv=cv,
        scoring=scoring
    )

    heart_cv_results.append({
        "Model": model_name,
        "Mean Accuracy": scores["test_accuracy"].mean(),
        "Mean Precision": scores["test_precision"].mean(),
        "Mean Recall": scores["test_recall"].mean(),
        "Mean F1 Score": scores["test_f1"].mean()
    })

heart_cv_df = pd.DataFrame(heart_cv_results)

print("=" * 70)
print("      HEART DISEASE - 5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 70)

display(heart_cv_df.round(4))
```

```python
def predict_diabetes(
    pregnancies,
    glucose,
    blood_pressure,
    skin_thickness,
    insulin,
    bmi,
    diabetes_pedigree,
    age
):
    # Create input DataFrame with original feature names
    input_data = pd.DataFrame([{
        "Pregnancies": pregnancies,
        "Glucose": glucose,
        "BloodPressure": blood_pressure,
        "SkinThickness": skin_thickness,
        "Insulin": insulin,
        "BMI": bmi,
        "DiabetesPedigreeFunction": diabetes_pedigree,
        "Age": age
    }])

    # Apply the same preprocessing used during training
    input_scaled = diabetes_scaler.transform(input_data)

    # Select the same features used during training
    input_selected = diabetes_selector.transform(input_scaled)

    # Use the best model selected in Step 6
    model = diabetes_models[best_diabetes_model_name]

    prediction = model.predict(input_selected)[0]
    probability = model.predict_proba(input_selected)[0][1]

    if prediction == 1:
        result = "⚠️ Diabetes Prediction: POSITIVE"
    else:
        result = "✅ Diabetes Prediction: NEGATIVE"

    return f"{result}\nProbability of Diabetes: {probability * 100:.2f}%"
```

```python
def predict_heart(
    age,
    sex,
    cp,
    trestbps,
    chol,
    fbs,
    restecg,
    thalach,
    exang,
    oldpeak,
    slope,
    ca,
    thal
):
    # Create input DataFrame with original feature names
    input_data = pd.DataFrame([{
        "age": age,
        "sex": sex,
        "cp": cp,
        "trestbps": trestbps,
        "chol": chol,
        "fbs": fbs,
        "restecg": restecg,
        "thalach": thalach,
        "exang": exang,
        "oldpeak": oldpeak,
        "slope": slope,
        "ca": ca,
        "thal": thal
    }])

    # Apply the same preprocessing used during training
    input_scaled = heart_scaler.transform(input_data)

    # Select the same features used during training
    input_selected = heart_selector.transform(input_scaled)

    # Use the best model selected in Step 6
    model = heart_models[best_heart_model_name]

    prediction = model.predict(input_selected)[0]
    probability = model.predict_proba(input_selected)[0][1]

    if prediction == 1:
        result = "⚠️ Heart Disease Prediction: POSITIVE"
    else:
        result = "✅ Heart Disease Prediction: NEGATIVE"

    return f"{result}\nProbability of Heart Disease: {probability * 100:.2f}%"
```

```python
with gr.Blocks(title="Disease Prediction System") as app:

    gr.Markdown(
        """
        # 🏥 Disease Prediction System

        This system uses Machine Learning to predict the possibility of
        **Diabetes** and **Heart Disease**.

        **Note:** This project is for educational/research purposes only
        and must not be used as a medical diagnosis.
        """
    )

    # ---------------- DIABETES ----------------

    with gr.Tab("Diabetes Prediction"):

        gr.Markdown("## Enter Patient Information")

        with gr.Row():
            pregnancies = gr.Number(
                label="Pregnancies",
                value=1
            )
            glucose = gr.Number(
                label="Glucose",
                value=120
            )

        with gr.Row():
            blood_pressure = gr.Number(
                label="Blood Pressure",
                value=70
            )
            skin_thickness = gr.Number(
                label="Skin Thickness",
                value=20
            )

        with gr.Row():
            insulin = gr.Number(
                label="Insulin",
                value=80
            )
            bmi = gr.Number(
                label="BMI",
                value=25
            )

        with gr.Row():
            diabetes_pedigree = gr.Number(
                label="Diabetes Pedigree Function",
                value=0.5
            )
            age = gr.Number(
                label="Age",
                value=30
            )

        diabetes_button = gr.Button(
            "Predict Diabetes"
        )

        diabetes_output = gr.Textbox(
            label="Prediction Result"
        )

        diabetes_button.click(
            fn=predict_diabetes,
            inputs=[
                pregnancies,
                glucose,
                blood_pressure,
                skin_thickness,
                insulin,
                bmi,
                diabetes_pedigree,
                age
            ],
            outputs=diabetes_output
        )

    # ---------------- HEART DISEASE ----------------

    with gr.Tab("Heart Disease Prediction"):

        gr.Markdown("## Enter Patient Information")

        with gr.Row():
            heart_age = gr.Number(label="Age", value=50)
            sex = gr.Number(label="Sex (0 = Female, 1 = Male)", value=1)

        with gr.Row():
            cp = gr.Number(label="Chest Pain Type (cp)", value=0)
            trestbps = gr.Number(label="Resting Blood Pressure", value=120)

        with gr.Row():
            chol = gr.Number(label="Cholesterol", value=200)
            fbs = gr.Number(label="Fasting Blood Sugar (fbs)", value=0)

        with gr.Row():
            restecg = gr.Number(label="Resting ECG (restecg)", value=0)
            thalach = gr.Number(label="Maximum Heart Rate", value=150)

        with gr.Row():
            exang = gr.Number(label="Exercise Induced Angina (exang)", value=0)
            oldpeak = gr.Number(label="ST Depression (oldpeak)", value=1.0)

        with gr.Row():
            slope = gr.Number(label="Slope", value=1)
            ca = gr.Number(label="Number of Major Vessels (ca)", value=0)

        thal = gr.Number(
            label="Thalassemia (thal)",
            value=2
        )

        heart_button = gr.Button(
            "Predict Heart Disease"
        )

        heart_output = gr.Textbox(
            label="Prediction Result"
        )

        heart_button.click(
            fn=predict_heart,
            inputs=[
                heart_age,
                sex,
                cp,
                trestbps,
                chol,
                fbs,
                restecg,
                thalach,
                exang,
                oldpeak,
                slope,
                ca,
                thal
            ],
            outputs=heart_output
        )


print("UI created successfully!")
```
